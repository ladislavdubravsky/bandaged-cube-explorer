# Optimal colored search

Exhaustively count unique reachable colored states by their shortest solution distance to solved. Suitable for small puzzles only!

Run with the **v2/.venv** kernel.

```python
import bce_v2 as c
from collections import Counter
from IPython.display import Markdown, display

bandage = [
    1, 1, 2,
    3, 3, 2,
    4, 4, 0,

    5, 5, 6,
    5, 5, 6,
    7, 8, 9,

    5, 5, 10,
    5, 5, 10,
    7, 8, 9,
]
```

```python
initial = c.State(bandage)
htm_graph = c.explore_colored(initial, metric="HTM")
qtm_graph = c.explore_colored(initial, metric="QTM")
assert htm_graph.complete and qtm_graph.complete

htm_distances = htm_graph.distances()
qtm_distances = qtm_graph.distances()
htm_counts = Counter(htm_distances)
qtm_counts = Counter(qtm_distances)
assert None not in htm_counts and None not in qtm_counts

shape_counts = Counter(state.shape for state in htm_graph)
display(Markdown(
    f"Both searches are complete: **{len(htm_graph):,} colored states** across "
    f"**{len(shape_counts):,} shapes**.\n\n"
    "| Metric | Colored states | Maximum distance from solved | Mean optimal distance |\n"
    "| --- | ---: | ---: | ---: |\n"
    f"| HTM | {len(htm_graph):,} | {max(htm_distances)} | {sum(htm_distances) / len(htm_distances):.4f} |\n"
    f"| QTM | {len(qtm_graph):,} | {max(qtm_distances)} | {sum(qtm_distances) / len(qtm_distances):.4f} |"
))
display(Markdown(
    "| Optimal distance | Colored states (HTM) | Colored states (QTM) |\n"
    "| ---: | ---: | ---: |\n"
    + "\n".join(
        f"| {distance} | {htm_counts[distance]:,} | {qtm_counts[distance]:,} |"
        for distance in range(max(max(htm_counts), max(qtm_counts)) + 1)
    )
    + f"\n| **Total** | **{sum(htm_counts.values()):,}** | **{sum(qtm_counts.values()):,}** |"
))
```
